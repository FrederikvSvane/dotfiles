Run all cells. You should see a static plot, then an animation rendered to `wave.mp4`.

```python
import numpy as np
import matplotlib.pyplot as plt

# Catppuccin Mocha so plots match the terminal
plt.rcParams.update({
    "figure.facecolor": "#1e1e2e", "axes.facecolor": "#1e1e2e",
    "axes.edgecolor": "#6c7086", "axes.labelcolor": "#cdd6f4",
    "xtick.color": "#cdd6f4", "ytick.color": "#cdd6f4",
    "text.color": "#cdd6f4", "grid.color": "#313244",
    "axes.prop_cycle": plt.cycler(color=["#89b4fa", "#f38ba8", "#a6e3a1", "#f9e2af"]),
})
print("numpy", np.__version__)
```

## 1. Static plot

```python
x = np.linspace(0, 4 * np.pi, 400)
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.set_title("Static plot")
ax.grid(True)
ax.legend()
plt.show()
```

```python
from matplotlib.animation import FuncAnimation, FFMpegWriter, PillowWriter

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.set_xlim(0, 4 * np.pi)
ax.set_ylim(-1.2, 1.2)
ax.grid(True)
(line,) = ax.plot([], [], lw=2)

def update(i):
    line.set_data(x, np.sin(x - i / 10 * np.pi))
    return (line,)

anim = FuncAnimation(fig, update, frames=60, blit=True)
try:
    anim.save("wave.mp4", writer=FFMpegWriter(fps=30))
except Exception as e:  # ffmpeg missing or broken: the gif below still works
    print("mp4 skipped:", type(e).__name__)
anim.save("wave.gif", writer=PillowWriter(fps=30))
plt.close(fig)
print("wrote wave.gif")
```

### Watching it in the terminal
Terminals cannot play video inside a notebook output, so view the files directly:

- `kitten icat --loop=-1 --hold wave.gif` animates the gif (also works inside tmux and over ssh)
- `mpv --vo=kitty wave.mp4` plays the video (needs mpv; not inside tmux)
- in nvim, put the cursor on `wave.gif` and press `<localleader>mp`
